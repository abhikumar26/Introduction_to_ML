```python
import pandas as pd
import numpy as np
```

```python
data = {
    "age": [
        20, 19, 18, 22, 21,
        23, 24, 20, 19, 22,
        25, 26, 23, 27, 21
    ],

    "salary": [
        25000, 13000, 10000, 15000, 18000,
        20000, 17000, 10600, 10400, 19000,
        22000, 20500, 20100, 24000, 23000
    ]

}

df = pd.DataFrame(data)
df
```

```python
from sklearn.preprocessing import MinMaxScaler
import numpy as np
```

```python
scaler = MinMaxScaler()

val = scaler.fit_transform(df)
val = np.round(val, 2)
val

dff = pd.DataFrame(val, columns=["age", "salary"])
dff
```

```python
from sklearn.cluster import KMeans
```

```python
km = KMeans(n_clusters=3 , random_state=40)
km.fit_predict(dff)
```

```python
sse_data = []
for i in range(1, 14):
    km = KMeans(n_clusters=i, random_state=40)
    km.fit_predict(dff)
    sse_data.append(km.inertia_)
np.round(sse_data, 3)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.Figure(figsize=(8,6))
plt.plot
plt.plot(sse_data, marker='o', color="red")
plt.xlabel("K")
plt.ylabel("Error")
plt.show()
```
