```python
import numpy as np
import pandas as pd
```

```python
data = np.array([1,2,3,3,445,6,7,7,8,8,8,86,6,6,6,6888,1])
q1 = np.percentile(data,25)
q2 = np.percentile(data,50)
q3 = np.percentile(data,75)
q1
```

```python
df = pd.DataFrame({"Value":data})
df
```

```python
df.describe()
```

```python
q1 = df.quantile(0.25)
q2 = df.quantile(0.50)
q3 = df.quantile(0.75)
q3
```

```python
IQR = q3-q1 # IQR formula
IQR
```

```python
lower = q1-(IQR*0.25)
upper = q3+(IQR*0.75)
upper
```
