```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("insurance.csv")
df.drop(["region","smoker"],axis=1,inplace=True)
df
```

```python
q1= df[["bmi"]].quantile(0.25)
q3 = df[["bmi"]].quantile(0.75)
IQR = q3-q1
lower = q1 - IQR * IQR
upper = q3 + IQR * IQR
outliers = np.where(df[(df[["bmi"]] < lower) | (df[["bmi"]] > upper)])
outliers
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("insurance.csv")
df = df.drop(columns=["region", "smoker"])

numeric_cols = ["age", "bmi", "children"]

plt.figure(figsize=(12, 6))
bp = plt.boxplot([df[c] for c in numeric_cols], patch_artist=True)

plt.xticks(range(1, len(numeric_cols) + 1), numeric_cols)
plt.title("Boxplots of Age, BMI, Children, and Charges")
plt.ylabel("Values")
plt.show()
```
